```python
#Step 1: Import Libraries
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import f_oneway

sns.set_style("whitegrid")

print("Libraries imported successfully.")
```

```python
#Step 2: Load All 9 Datasets from GitHub

import pandas as pd

url = 'https://raw.githubusercontent.com/RavnaSrikanth/Ravna---Capstone-Project/refs/heads/main/Categories.csv'
base_url = "https://raw.githubusercontent.com/RavnaSrikanth/Ravna---Capstone-Project/main/"

customers = pd.read_csv(base_url + "Customers.csv")
orders = pd.read_csv(base_url + "Orders.csv")
order_items = pd.read_csv(base_url + "Order%20Items.csv")
products = pd.read_csv(base_url + "Products.csv")
categories = pd.read_csv(base_url + "Categories.csv")
departments = pd.read_csv(base_url + "Departments.csv")
shipping = pd.read_csv(base_url + "Shipping.csv")
geography = pd.read_csv(base_url + "Geography.csv")
payment_types = pd.read_csv(base_url + "Payment%20Types.csv")

print("All datasets loaded successfully.")
```

```python
#Step 3: Explore Dataset Sizes

datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Products": products,
    "Categories": categories,
    "Departments": departments,
    "Shipping": shipping,
    "Geography": geography,
    "Payment Types": payment_types
}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

```python
#Step 4: Check Column Names

for name, df in datasets.items():
    print("\n" + "="*60)
    print(name)
    print("="*60)

    print(df.columns.tolist())
```

```python
#Step 5: Check Missing Values

for name, df in datasets.items():
    print("\n" + "="*60)
    print(name)
    print("="*60)

    print(df.isnull().sum())
```

```python
#Step 6: Check Duplicate Records
for name, df in datasets.items():
    print(
        f"{name}: {df.duplicated().sum()} duplicate rows"
    )
```

```python
#Create Shipping Delay

orders["Shipping Delay"] = (
orders["Days for shipping (real)"]
- orders["Days for shipment (scheduled)"]
)
orders[[
"Days for shipping (real)",
"Days for shipment (scheduled)",
"Shipping Delay"
]].head()
```

```python
#First Visualization

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,6))

sns.barplot(
    data=orders,
    x='Order Region',
    y='Shipping Delay',
    estimator='mean'
)

plt.xticks(rotation=45)
plt.title('Average Shipping Delay by Region')
plt.show()
```
